```python
import pandas as pd
import numpy as np
```

## Daily weather 

```python
df_weather = pd.read_csv('../../data/intermediate/weather_daily.csv')
df_weather
```

```python
# Handle temperature outliers
temp_outliers = df_weather.temperature.nlargest(2).index
df_weather.loc[temp_outliers, 'temperature'] = np.nan

# Handle humidity outliers by clipping to 100
df_weather['humidity'] = df_weather['humidity'].clip(upper=100)

# Handle precipitation outliers
precip_outliers = df_weather.precipation.nlargest(50).index
df_weather.loc[precip_outliers, 'precipation'] = np.nan

# Convert precipitation values less than 0.01 to zero
df_weather.loc[df_weather.precipation < 0.01, 'precipation'] = 0
```

```python
print(df_weather.describe())
```

```python
df_weather.to_csv('../../data/processed/weather_daily.csv', index=False)
```

## Incidents weather

```python
df_incidents = pd.read_csv('../../data/intermediate/incidents.csv')
df_incidents
```

```python
df_incidents.describe()
```

```python
# Find extreme temperature values
highest_temps = df_incidents.temperature.nlargest(49)
lowest_temps = df_incidents.temperature.nsmallest(42)

# Combine highest and lowest temps into one Series of outliers
temp_outliers = pd.concat([highest_temps, lowest_temps])

# Set these extreme values to NaN
df_incidents.loc[temp_outliers.index, 'temperature'] = np.nan
```

```python
print(df_incidents[['temperature', 'humidity', 'precipation', 'precipation_24h']].describe())
```

```python
df_incidents.to_csv('../../data/processed/incidents.csv', index=False)
```
